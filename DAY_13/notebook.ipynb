```python
from pathlib import Path

import torchvision
from torchvision.datasets import CIFAR10

data_dir = Path("data")
data_dir.mkdir(exist_ok=True)

train_data = CIFAR10(data_dir, train=True, download=True, transform=torchvision.transforms.ToTensor())

print(len(train_data))
```

### EDA

```python
print(train_data.classes)
print(train_data[0])
```

```python
from collections import Counter

print(train_data[0][0].shape)
print(Counter(train_data.targets))
```

```python
i = train_data.targets.index(1)
image, label = train_data[i]
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 5, figsize=(10,4))
axes_flat = axes.flatten()

for i, class_name in enumerate(train_data.classes):
    idx = train_data.targets.index(i)
    axes_flat[i].imshow(train_data[idx][0].permute(1, 2, 0))
    axes_flat[i].set_title(class_name)
```
